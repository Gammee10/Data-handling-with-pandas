```python
df = pd.read_csv("Titanic-Dataset.csv")

df.head()
df.info()
df.describe()
```

## Step 2: Data Cleaning

## Objective
Handle missing values and remove unnecessary data.

## Actions Taken
- Fill missing values in Age using median
- Fill missing values in Embarked using mode
- Drop Cabin column due to many missing values
- Remove duplicate rows

```python
df["Age"] = df["Age"].fillna(df["Age"].median())
df["Embarked"] = df["Embarked"].fillna(df["Embarked"].mode()[0])
df.drop("Cabin", axis=1, inplace=True)
df.drop_duplicates(inplace=True)
```

## Step 3: Data Analysis

## Objective
Analyze survival patterns using groupby operations.

## Analysis Performed
- Survival rate by gender
- Survival rate by passenger class
- Average age per class
- Survival rate by age group

```python
# Survival by gender
df.groupby("Sex")["Survived"].mean()

# Survival by class
df.groupby("Pclass")["Survived"].mean()

# Average age per class
df.groupby("Pclass")["Age"].mean()

# Create age groups
df["AgeGroup"] = pd.cut(df["Age"], bins=[0,12,18,35,60,100],
                       labels=["Child","Teen","Young Adult","Adult","Senior"])

# Survival by age group
df.groupby("AgeGroup")["Survived"].mean()
```

## Step 4: Data Filtering

## Objective
Extract specific groups of passengers for deeper insights.

## Filters Applied
- Female passengers who survived
- Children who survived
- First-class passengers who survived

```python
# Female survivors
df[(df["Sex"] == "female") & (df["Survived"] == 1)]

# Children survivors
df[(df["Age"] < 12) & (df["Survived"] == 1)]

# First class survivors
df[(df["Pclass"] == 1) & (df["Survived"] == 1)]
```

## Step 5: Insights

## Key Findings

- Females had a significantly higher survival rate than males.
- Passenger class strongly influenced survival chances, with 1st class having the highest survival rate.
- Children showed relatively higher survival rates, supporting the "women and children first" principle.
- The highest survival probability was observed among female passengers in 1st class.
